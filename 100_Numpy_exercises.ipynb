#### 22. Normalize a 5x5 random matrix (★☆☆)

````python
# Min-Max Normalization
matrix_2=np.arange(1,26).reshape(5,5)
matrix_2=(matrix_2-np.min(matrix_2))/(np.max(matrix_2)-np.min(matrix_2))
print(matrix_2)

#Z-Score Standardization
matrix_3=np.arange(50,75).reshape(5,5)
matrix_3=(matrix_3-np.mean(matrix_3))/(np.std(matrix_3))
print(matrix_3)
````

#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆)

````python
A=np.array([1,2,3,4,5],dtype='float64')
B=np.array([6,7,8,9,10],dtype='float64')

np.add(A,B,out=B)
np.divide(A,2,out=A)
np.negative(A,out=A)
np.multiply(A,B,out=A)
answer(35)
````

#### 36. Extract the integer part of a random array of positive numbers using 4 different methods (★★☆)

````python
random_array=np.random.random(10)*10

print(random_array)

#First Method
print(np.floor(random_array))

#Second Method
print(random_array.astype('int'))

#Third Method
print(np.trunc(random_array))

#Fourth Method
print(random_array//1)

#Fifth Method
print(random_array-random_array%1)

````
